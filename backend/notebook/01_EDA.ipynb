```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Style cho chart đẹp hơn
sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams['figure.figsize'] = (10, 5)
```

```python
df = pd.read_csv('/content/data.csv')

print(f"Shape: {df.shape}")
df.head()
```

```python
print("=== THÔNG TIN CỘT ===")
print(df.info())

print("\n=== MISSING VALUES ===")
missing = df.isnull().sum()
print(missing[missing > 0] if missing.sum() > 0 else "Không có missing values ✓")

print("\n=== DUPLICATE ROWS ===")
print(f"Số dòng trùng: {df.duplicated().sum()}")

print("\n=== THỐNG KÊ MÔ TẢ ===")
df.describe()
```

```python
print("=== PHÂN PHỐI GIÁ ===")
print(f"Min:    ${df['price'].min():>12,.0f}")
print(f"Max:    ${df['price'].max():>12,.0f}")
print(f"Mean:   ${df['price'].mean():>12,.0f}")
print(f"Median: ${df['price'].median():>12,.0f}")
print(f"Skewness: {df['price'].skew():.3f}")

# Vẽ phân phối giá gốc
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].hist(df['price'], bins=80, color='steelblue', edgecolor='none', alpha=0.8)
axes[0].set_title('Phân phối giá gốc')
axes[0].set_xlabel('Price (USD)')
axes[0].set_ylabel('Count')
axes[0].xaxis.set_major_formatter(
    plt.FuncFormatter(lambda x, _: f'${x/1e6:.1f}M'))

axes[1].hist(np.log(df['price'][df['price'] > 0]),
             bins=60, color='teal', edgecolor='none', alpha=0.8)
axes[1].set_title('Phân phối log(price) — gần chuẩn hơn')
axes[1].set_xlabel('log(Price)')
axes[1].set_ylabel('Count')

plt.tight_layout()
plt.show()
```

```python
# Xem các percentile để chọn ngưỡng cắt hợp lý
percentiles = [1, 5, 10, 25, 50, 75, 90, 95, 99, 100]
for p in percentiles:
    val = df['price'].quantile(p / 100)
    print(f"  P{p:3d}: ${val:>12,.0f}")

print(f"\nGiá = $0: {(df['price'] == 0).sum()} rows")
print(f"Giá < $50K: {(df['price'] < 50000).sum()} rows")
print(f"Giá > $2M:  {(df['price'] > 2_000_000).sum()} rows")

# Top 10 nhà đắt nhất
print("\n=== TOP 10 ĐẮT NHẤT ===")
print(df.nlargest(10, 'price')[
    ['price', 'sqft_living', 'bedrooms', 'bathrooms', 'city']
].to_string())
```

```python
num_features = ['sqft_living', 'sqft_above', 'sqft_basement',
                'bedrooms', 'bathrooms', 'floors',
                'yr_built', 'yr_renovated']

fig, axes = plt.subplots(2, 4, figsize=(18, 8))
axes = axes.flatten()

for i, col in enumerate(num_features):
    axes[i].hist(df[col], bins=50, color='steelblue',
                 edgecolor='none', alpha=0.8)
    axes[i].set_title(col)
    axes[i].set_ylabel('Count')

plt.suptitle('Phân phối các numeric features', fontsize=14, y=1.01)
plt.tight_layout()
plt.show()
```

```python
# Encode các cột cần thiết trước khi tính correlation
df_corr = df.copy()
df_corr['was_renovated'] = (df_corr['yr_renovated'] > 0).astype(int)

feature_cols = ['price', 'sqft_living', 'sqft_above', 'sqft_basement',
                'bedrooms', 'bathrooms', 'floors', 'waterfront',
                'view', 'condition', 'yr_built', 'was_renovated']

corr_matrix = df_corr[feature_cols].corr()

plt.figure(figsize=(12, 9))
mask = np.triu(np.ones_like(corr_matrix, dtype=bool))  # chỉ hiện nửa dưới
sns.heatmap(
    corr_matrix,
    mask=mask,
    annot=True,
    fmt='.2f',
    cmap='RdYlGn',
    center=0,
    square=True,
    linewidths=0.5,
    cbar_kws={'shrink': 0.8}
)
plt.title('Correlation matrix', fontsize=14)
plt.tight_layout()
plt.show()

# In riêng correlation với price để dễ đọc
print("\n=== CORRELATION VỚI PRICE (sắp xếp) ===")
print(corr_matrix['price'].sort_values(ascending=False).to_string())
```

```python
fig, axes = plt.subplots(2, 3, figsize=(16, 10))

# Giá theo số phòng ngủ
df_clean_viz = df[(df['price'] > 50000) & (df['price'] < 2_000_000)].copy()
df_clean_viz['was_renovated'] = (df_clean_viz['yr_renovated'] > 0).astype(int)

# 1. Bedrooms
bed_price = df_clean_viz.groupby('bedrooms')['price'].median()
bed_price = bed_price.loc[1:7]
axes[0, 0].bar(bed_price.index, bed_price.values / 1000,
               color='steelblue', alpha=0.85)
axes[0, 0].set_title('Median giá theo số phòng ngủ')
axes[0, 0].set_xlabel('Bedrooms')
axes[0, 0].set_ylabel('Median price ($K)')

# 2. View rating
view_price = df_clean_viz.groupby('view')['price'].median()
axes[0, 1].bar(view_price.index, view_price.values / 1000,
               color='teal', alpha=0.85)
axes[0, 1].set_title('Median giá theo View rating (0–4)')
axes[0, 1].set_xlabel('View')
axes[0, 1].set_ylabel('Median price ($K)')

# 3. Condition
cond_price = df_clean_viz.groupby('condition')['price'].median()
axes[0, 2].bar(cond_price.index, cond_price.values / 1000,
               color='coral', alpha=0.85)
axes[0, 2].set_title('Median giá theo Condition (1–5)')
axes[0, 2].set_xlabel('Condition')
axes[0, 2].set_ylabel('Median price ($K)')

# 4. Sqft vs Price scatter
axes[1, 0].scatter(df_clean_viz['sqft_living'],
                   df_clean_viz['price'] / 1000,
                   alpha=0.15, s=8, color='steelblue')
axes[1, 0].set_title('sqft_living vs Price')
axes[1, 0].set_xlabel('sqft_living')
axes[1, 0].set_ylabel('Price ($K)')

# 5. Waterfront
wf = df_clean_viz.groupby('waterfront')['price'].median() / 1000
axes[1, 1].bar(['Không ven hồ', 'Ven hồ'], wf.values,
               color=['steelblue', 'coral'], alpha=0.85)
axes[1, 1].set_title('Median giá: Waterfront')
axes[1, 1].set_ylabel('Median price ($K)')

# 6. Yr_built theo thập niên
df_clean_viz['decade'] = (df_clean_viz['yr_built'] // 10 * 10)
decade_price = df_clean_viz.groupby('decade')['price'].median() / 1000
axes[1, 2].plot(decade_price.index, decade_price.values,
                marker='o', color='teal', linewidth=2)
axes[1, 2].set_title('Median giá theo thập niên xây dựng')
axes[1, 2].set_xlabel('Decade built')
axes[1, 2].set_ylabel('Median price ($K)')

plt.suptitle('Phân tích giá theo các yếu tố chính', fontsize=14, y=1.01)
plt.tight_layout()
plt.show()
```

```python
city_stats = (
    df_clean_viz.groupby('city')['price']
    .agg(['median', 'count'])
    .query('count >= 20')          # bỏ city ít hơn 20 records
    .sort_values('median', ascending=True)
)

plt.figure(figsize=(10, 8))
plt.barh(city_stats.index, city_stats['median'] / 1000,
         color='steelblue', alpha=0.85)
plt.xlabel('Median price ($K)')
plt.title('Median giá nhà theo thành phố (≥20 giao dịch)')
plt.tight_layout()
plt.show()

print(city_stats.sort_values('median', ascending=False).to_string())
```
